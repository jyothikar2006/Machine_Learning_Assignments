```python
lst = [2,7,4,1,3,6]
counter = 0
for i in range(len(lst)):
    for j in range(i+1, len(lst)):
        pairSum = lst[i] + lst[j]
        if pairSum == 10:
            counter+=1
print("Number of pairs: ", counter)
```

```python
lst = list(map(int, input("Enter number:").split()))
if len(lst) < 3:
    print("Range determination not pissible.")
else:
    minimum = lst[0]
    maximum = lst[0]
    for i in range(1, len(lst)):
        if lst[i] < minimum:
            minimum = lst[i]
        if lst[i] > maximum:
            maximum = lst[i]
        rg = maximum - minimum
print("Range:", rg)
```

```python
rows = int(input("Enter rows:"))
cols = int(input("Enter columns:"))
if rows == cols:
    lst = []
    for i in range(rows):
        input_lst = (list(map(int, input("Enter for a single row:").split())))
        lst.append(input_lst)
    m = int(input("Enter a postive number:"))
    res = [row[:] for row in lst]
    for _ in range(m-1):
        temp = [[0 for _ in range(cols)] for _ in range(rows)]
        for i in range(rows):
            for j in range(cols):
                for k in range(cols):
                        temp[i][j] += res[i][k] * lst[k][j]
        res = temp
    print(res)
```

```python
string = input("Enter a string:")
occurrence = {}
for ch in string:
    if ch not in occurrence:
        occurrence[ch] = 1
    else:
        occurrence[ch] += 1

max_val = 0
for x in occurrence.items():
    if x[1] > max_val:
        max_val = x[1]
        max_ch = x[0]
print("character:", max_ch)
print("occurrrence:", max_val)
```

```python
import random
numbers = []
for _ in range(25):
    numbers.append(random.randint(1, 10))
print(numbers)

def mean(nums):
    sum = 0
    for i in nums:
        sum += i
    avg = sum / len(nums)
    print("Mean:", avg)

def median(nums):
    nums.sort()
    print("Sorted list:", nums)
    index = len(nums) // 2
    print("Median:", nums[index])

def mode(nums):
    occurrence = {}
    for n in nums:
        if n not in occurrence:
            occurrence[n] = 1
        else:
            occurrence[n] += 1
    print(occurrence)

print(mean(numbers))
print(median(numbers))
print("Mode:", mode(numbers))
```
